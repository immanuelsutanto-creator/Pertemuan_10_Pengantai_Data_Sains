```python
# cell 1 - Import Library
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

cell 2: Data Sintetis

```python
np.random.seed(2026)
n_samples = 100

X1 = np.random.rand(n_samples) * 2
X2 = np.random.rand(n_samples) * 3

noise = np.random.rand(n_samples) * 1.5

#pengaruh linear: Y = 5 + 3.5*X1 - 2 *X2 + Noise
Y = 5 + 3.5 * X1 - 2 * X2 + noise

#Menggabungkan menjadi data frame
df = pd.DataFrame({'X1': X1, 'X2': X2, 'Y': Y})
display(df.head())
```

cell 3: Model Fit

```python
X = df[['X1', 'X2']]
y = df['Y']

#membagi training 80% testing 20%
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2026)

model = LinearRegression()
model.fit(X_train, y_train)

#Melihat intercept dan koefisien
print(f"Intercept (konstanta): {model.intercept_:.4f}")
print(f"Koefisien X1: {model.coef_[0]:.4f}")
print(f"koefisien X2: {model.coef_[1]:.4f}")

# Y = 5.92 + 3,42 X1 - 2,10 X2
```

cell 4: Prediksi dan Matrix

```python
y_pred = model.predict(X_test)

#Menghitung MSE dan R2 (Matrix)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

#print Matrix
print(f"Mean Squared Error (MSE): {mse:.4f}")
print(f"R-squared (R2): {r2:.4f}")
```

cell 5: Visualisasi

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Prediksi pada data testing
y_pred = model.predict(X_test)

plt.figure(figsize=(8, 6))

sns.scatterplot(x=y_test, y=y_pred, color='blue', alpha=0.7)
plt.plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], color='red', linestyle='--', linewidth=2)

plt.title("Nilai Aktual vs Prediksi")
plt.xlabel("Nilai Aktual (Y)")
plt.ylabel("Prediksi (Y_topi)")
plt.grid(True)
plt.show()
```
